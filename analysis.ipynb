## Cell 1: Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## Cell 2: Load Data

```python
# Load the Aadhaar enrollment data
df = pd.read_csv('data/aadhaar_enrollment.csv')

print(f"Total rows: {len(df)}")
print(f"Total columns: {len(df.columns)}")

# Display first few rows
df.head()
```

## Cell 3: Basic Statistics

```python
# Display basic statistics
df.describe()
```

## Cell 4: Check for Missing Values

```python
# Check for null values
print("Missing values per column:")
df.isnull().sum()
```

## Cell 5: State-wise Analysis

```python
# Top 10 states by enrollment
state_counts = df['state'].value_counts().head(10)
print("Top 10 States by Enrollment:")
print(state_counts)

# Visualization
plt.figure(figsize=(12, 6))
state_counts.plot(kind='barh', color='skyblue')
plt.title('Top 10 States by Aadhaar Enrollment')
plt.xlabel('Number of Enrollments')
plt.tight_layout()
plt.show()
```

## Cell 6: Gender Distribution

```python
# Gender distribution
gender_counts = df['gender'].value_counts()
print("Gender Distribution:")
print(gender_counts)

# Pie chart
plt.figure(figsize=(8, 8))
plt.pie(gender_counts.values, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Gender Distribution in Aadhaar Enrollment')
plt.axis('equal')
plt.show()
```

## Cell 7: Age Analysis

```python
# Age statistics
print("Age Statistics:")
print(df['age'].describe())

# Age distribution histogram
plt.figure(figsize=(12, 6))
plt.hist(df['age'], bins=30, color='green', alpha=0.7, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.grid(True, alpha=0.3)
plt.show()
```

## Cell 8: Age Group Analysis

```python
# Create age groups
age_bins = [0, 18, 30, 45, 60, 100]
age_labels = ['0-17', '18-29', '30-44', '45-59', '60+']
df['age_group'] = pd.cut(df['age'], bins=age_bins, labels=age_labels)

age_group_counts = df['age_group'].value_counts().sort_index()
print("Enrollment by Age Group:")
print(age_group_counts)

# Bar chart
plt.figure(figsize=(10, 6))
age_group_counts.plot(kind='bar', color='orange')
plt.title('Enrollment by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Cell 9: Key Insights Summary

```python
# Summary insights
print("="*60)
print("KEY INSIGHTS SUMMARY")
print("="*60)

total_enrollments = len(df)
unique_states = df['state'].nunique()
top_state = df['state'].value_counts().index[0]
top_state_count = df['state'].value_counts().values[0]

print(f"\n✓ Total Enrollments: {total_enrollments:,}")
print(f"✓ Unique States: {unique_states}")
print(f"✓ Top State: {top_state} ({top_state_count:,})")
print(f"✓ Average Age: {df['age'].mean():.1f} years")
print(f"✓ Analysis Date: {pd.Timestamp.now().strftime('%Y-%m-%d')}")
```
